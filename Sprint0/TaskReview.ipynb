```
Plant myPlant = new Plant();
Flower myFlower = new Flower();
Tree myTree = new Tree();

myPlant.Name = "Горох";
myPlant.Kind = "Растение";
myPlant.Height = 1;
myPlant.Blossom();
Console.WriteLine($"Высота {myPlant.Kind} {myPlant.Name} - {myPlant.Height}");
myPlant.Growing(3);
Console.WriteLine($"Высота {myPlant.Kind} {myPlant.Name} - {myPlant.Height}");

Console.WriteLine($"\n==========================================\n");

myFlower.Name = "Флауи";
myFlower.Kind = "Цветок";
myFlower.Color = "Желтый";
myFlower.Height = 1;
myFlower.Blossom();
Console.WriteLine($"Высота {myFlower.Kind} {myFlower.Name} - {myFlower.Height}");
myFlower.Growing(3);
Console.WriteLine($"Высота {myFlower.Kind} {myFlower.Name} - {myFlower.Height}");
myFlower.Wither();

Console.WriteLine($"\n==========================================\n");

myTree.Name = "Грут";
myTree.Kind = "Дерево";
myTree.Age = 49;
myFlower.Height = 20;
myTree.Blossom();
myTree.Growing(3);
Console.WriteLine($"Возраст {myTree.Kind} {myTree.Name} - {myTree.Age}");
myTree.ToGrow(23);

class Plant
{
    public string Name {get; set;}
    public string Kind {get; set;}
    public float Height {get; set;}

    public virtual void Blossom()
    {
        Console.WriteLine($"Растение {Name} цветет");
    }

    public virtual void Growing(float height)
    {
        Height += height;
    }
}

class Flower : Plant
{
    public string Color {get; set;}
    public int CountOfPetals {get; set;}

    public override void Blossom()
    {
        Console.WriteLine($"{Kind} {Name} цветет");
    }

    public void Wither()
    {
        Console.WriteLine($"{Kind} {Name} начинает увядать");
    }
}

class Tree : Plant
{
    public int Age {get; set;}
    public string LeavesType {get; set;}

    public override void Blossom()
    {
        Console.WriteLine($"{Kind} {Name} цветёт");
    }

    public void ToGrow(int years)
    {
        Age += years;
        Console.WriteLine($"Новый возраст дерева {Name} - {Age}");
    }
}
```
