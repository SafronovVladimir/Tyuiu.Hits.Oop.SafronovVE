```
Product product1 = new Product();
product1.Name = "Молоко";
product1.Description = "Ничего натурального";
product1.Category = "Молочный продукт";
product1.Quantity = 1;
product1.Price = 69;
product1.Quantity += 1;
Console.WriteLine(product1.GetProductInfo());

Console.WriteLine("================================");

Seller seller1 = new Seller();
seller1.Name = "Ашот";
seller1.EmployeeId = "1";
seller1.ContactInfo = "+7888887777";
seller1.Salary = 100000;
seller1.AddProduct(product1);
Console.WriteLine(seller1.GetSellerInfo());

Console.WriteLine("================================");

Store store1 = new Store();
store1.StoreName = "Продуктовый";
store1.storeHours = "9:00-20:00";
store1.Location = "ул. Пушкина, д. 28";
store1.AddSeller(seller1);
Console.WriteLine(store1.GetStoreInfo());
Console.WriteLine("================================");
store1.ListProducts();

class Product
{
    private string _name;
    private decimal _price;
    private int _quantity;
    protected string category;
    internal string description;

    public string Name
    {
        get => _name;
        set
        {
            _name = value;
        }
    }

    public decimal Price
    {
        get => _price;
        set
        {
            _price = value;
        }
    }

    public int Quantity
    {
        get => _quantity;
        set
        {
            _quantity = value;
        }
    }

    public string Description
    {
        get => description;
        set
        {
            description = value;
        }
    }

    public string Category
    {
        get => category;
        set
        {
            category = value;
        }
    }

    public decimal GetTotalPrice()
    {
        return Price * Quantity;
    }

    public void UpdateQuantity(int amount)
    {
        Quantity = amount;
    }

    public string GetProductInfo()
    {
        return $"О товаре\nНазвание: {Name}\nОписание: {Description}\nЦена: {Price}";
    }
}

class Seller
{
    private string _name;
    private string _employeeId;
    public List<Product> productsList = new List<Product> {};
    protected decimal salary;
    internal string contactInfo;

    public string Name
    {
        get => _name;
        set
        {
            _name = value;
        }
    }
    public string EmployeeId
    {
        get => _employeeId;
        set
        {
            _employeeId = value;
        }
    }

    public decimal Salary
    {
        get => salary;
        set
        {
            salary = value;
        }
    }
    public string ContactInfo
    {
        get => contactInfo;
        set
        {
            contactInfo = value;
        }
    }

    public void AddProduct(Product product)
    {
        productsList.Add(product);
    }

    public void SellProduct(Product product, int quantity)
    {
        product.Quantity -= quantity;
    }

    public string GetSellerInfo()
    {
        return $"Информация о продавце\nИмя: {Name}\nЗарплата: {Salary}\nНомер телефона: {ContactInfo}";
    }
}

class Store
{
    private string _storeName;
    private string _location;
    private List<Seller> _sellersList = new List<Seller> {};
    public string storeHours;

    public string StoreName
    {
        get => _storeName;
        set
        {
            _storeName = value;
        }
    }
    public string Location
    {
        get => _location;
        set
        {
            _location = value;
        }
    }

    public void AddSeller(Seller seller)
    {
        _sellersList.Add(seller);
    }

    public void ListProducts()
    {
        Console.WriteLine($"Продукты магазина \"{StoreName}\"");
        for (int i = 0; i < _sellersList.Count; i++)
        {
            var seller = _sellersList[i];
            for (int j = 0; j < seller.productsList.Count; j++)
            {
                var sellerProduct = seller.productsList[j];
                Console.WriteLine($"Продукт: {sellerProduct.Name}");
                Console.WriteLine($"Количество: {sellerProduct.Quantity}");
            }
        }

    }

    public string GetStoreInfo()
    {
        return $"Информация о магазине\nНазвание: {StoreName}\nАдрес: {Location}\nЧасы работы: {storeHours}";
    }
}
```
