```
OnlineServiceSubscription music = new OnlineServiceSubscription(1, "Яндекс Музыка", 249, 32, 100, true);

StreamingSubscription stream = new StreamingSubscription(2, "Кинопоиск", 399, 30, true, 5);

VideoSubscription video = new VideoSubscription(3, "YouTube Premium", 199, 1080, true);

OnlineServiceSubscriptionForStaff staffMusic = new OnlineServiceSubscriptionForStaff(4, "Яндекс Музыка для персонала", 5, "IT-отдел", 20, true);

stream.TopUpTheBalance(5000);
video.TopUpTheBalance(500);

Subscription[] subscriptions =
{
    music,
    stream,
    video,
    staffMusic
};

foreach (Subscription subscription in subscriptions)
{
    Console.WriteLine("================================");
    subscription.GetSubscriptionDetails();
    Console.WriteLine($"Месячная стоимость: {subscription.CalculateMonthlyCost()} руб.");
}

Console.WriteLine("================================");

stream.AddStream();
stream.AddStream();
stream.GetSubscriptionDetails();

Console.WriteLine("================================");

stream.ExtendSubscription(12);
stream.GetSubscriptionDetails();

Console.WriteLine("================================");

video.ChangeVideoQuality(720);
video.GetSubscriptionDetails();

Console.WriteLine("================================");

Console.WriteLine("Баланс до перевода:");
Console.WriteLine($"Кинопоиск: {stream.Balance} руб.");
Console.WriteLine($"YouTube Premium: {video.Balance} руб.");

stream.TransferBalance(video, 500);

Console.WriteLine("Баланс после перевода:");
Console.WriteLine($"Кинопоиск: {stream.Balance} руб.");
Console.WriteLine($"YouTube Premium: {video.Balance} руб.");
public interface ISubscription
{
    int SubscriptionId {get;}
    string ServiceName {get;}
    int Cost {get;}
    int Balance {get;}

    void TopUpTheBalance(int money);
    int CalculateMonthlyCost();
    void ExtendSubscription(int months);
    void GetSubscriptionDetails();
}

public interface IOnlineServiceSubscription
{
    int MaxUsers {get;}
    int CurrentUsers {get;}
    int StorageGb {get; set;}
    bool FamilyAccess {get; set;}
}

public interface IStreamingSubscription
{
    int MaxStreams {get;}
    int CurrentStreams {get;}
    bool CanRecord { get; set; }
    int DeviceLimit { get; set; }
    void AddStream();
    void StopStream();
    void SetMaxStreams(int maxStreams);
}

public interface IVideoSubscription
{
    int VideoQuality{get;}
    int FinalCost{get;}
    bool OfflineDownload {get; set;}
    void ChangeVideoQuality(int targetQuality);
}

class Subscription : ISubscription
{
    private int _subscriptionId;
    private string _serviceName;
    private int _cost;
    private int _balance;
    public int SubscriptionId
    {
        get => _subscriptionId;
        protected set
        {
            _subscriptionId = value;
        }
    }
    public string ServiceName
    {
        get => _serviceName;
        protected set
        {
            _serviceName = value;
        }
    }
    public int Cost
    {
        get => _cost;
        protected set
        {
            _cost = value;
        }
    }
    public int Balance
    {
        get => _balance;
        protected set
        {
            _balance = value;
        }
    }

    public Subscription(int subscriptionId, string serviceName, int cost)
    {
        SubscriptionId = subscriptionId;
        ServiceName = serviceName;
        Cost = cost;
    }
    
    public void TransferBalance(ISubscription subscription, int money)
    {
        if (Balance >= money)
        {
            Balance -= money;
            subscription.TopUpTheBalance(money);
        }
    }

    public DateTime EndDate {get; protected set;} = DateTime.Now.AddMonths(1);

    public void TopUpTheBalance(int money)
    {
        Balance += money;
    }
    
    public virtual int CalculateMonthlyCost()
    {
        return Cost;
    }
    public virtual void ExtendSubscription(int months)
    {
        int monthlyCost = CalculateMonthlyCost();
        int totalCost = monthlyCost * months;
        if (Balance < totalCost)
        {
            Console.WriteLine("Недостаточно средств для продления подписки");
            return;
        }
        EndDate = EndDate.AddMonths(months);
        Balance -= totalCost;
        Console.WriteLine("Подписка успешно продлена");
    }
    public virtual void GetSubscriptionDetails()
    {
        Console.WriteLine("Детали подписки");
        Console.WriteLine($"Id подписки: {SubscriptionId}");
        Console.WriteLine($"Название услуги: {ServiceName}");
        Console.WriteLine($"Дата окончания: {EndDate}");
        Console.WriteLine($"Средств на балансе: {Balance} руб.");
    }
}

class OnlineServiceSubscription : Subscription, IOnlineServiceSubscription
{
    private int _maxUsers = 32;
    private int _currentUsers;
    public int StorageGb {get; set;}
    public bool FamilyAccess {get; set;}
    public int MaxUsers
    {
        get => _maxUsers;
        set
        {
            _maxUsers = value;
        }
    }
    public int CurrentUsers
    {
        get => _currentUsers;
        set
        {
            _currentUsers = value;
        }
    }

    private string IsFamilyAccess => FamilyAccess ? "Включен" : "Не включен";

    public OnlineServiceSubscription(int subscriptionId, string serviceName, int cost, int currentUsers, int storageGb, bool familyAccess) : base(subscriptionId, serviceName, cost)
    {
        CurrentUsers = currentUsers;
        StorageGb = storageGb;
        FamilyAccess = familyAccess;
    }

    public override int CalculateMonthlyCost()
    {
        if (CurrentUsers == 0) return Cost;
        int dynamicCost = Convert.ToInt32(Cost * Math.Pow(CurrentUsers, -0.15));
        return dynamicCost;
    }

    public override void GetSubscriptionDetails()
    {
        base.GetSubscriptionDetails();
        Console.WriteLine($"Объём хранилища: {StorageGb} ГБ");
        Console.WriteLine($"Семейный доступ: {IsFamilyAccess}");
    }
}

class StreamingSubscription : Subscription, IStreamingSubscription
{
    private int _maxStreams = 20;
    private int _currentStreams = 0;
    public bool CanRecord { get; set; }
    public int DeviceLimit { get; set; }
    public int MaxStreams
    {
        get => _maxStreams;
        private set
        {
            _maxStreams = value;
        }
    }
    public int CurrentStreams
    {
        get => _currentStreams;
        private set
        {
            _currentStreams = value;
        }
    }

    public StreamingSubscription(int subscriptionId, string serviceName, int cost, int maxStreams, bool canRecord, int deviceLimit) : base(subscriptionId, serviceName, cost)
    {
        MaxStreams = maxStreams;
        CanRecord = canRecord;
        DeviceLimit = deviceLimit;
    }

    private string IsCanRecord => CanRecord ? "Включена" : "Не включена";

    public override void ExtendSubscription(int months)
    {
        if (months > 12)
        {
            Console.WriteLine("Подписку нельзя продлить больше, чем на 12 месяцев");
            return;
        }
        
        if (Balance < Convert.ToInt32(Cost * months * Math.Pow(months, -0.11)))
        {
            Console.WriteLine("Недостаточно средств для продления подписки");
            return;
        }

        EndDate = EndDate.AddMonths(months);
        Balance -= Convert.ToInt32(Cost * months * Math.Pow(months, -0.11));
        Console.WriteLine("Подписка успешно продлена");
    }

    public void AddStream()
    {
        if (CurrentStreams < MaxStreams)
        {
            CurrentStreams += 1;
            Console.WriteLine("Поток начат");
        }
        else
        {
            Console.WriteLine("В данный момент превышено максимальное количество одновременных потоков");
        }
    }

    public void StopStream()
    {
        if (CurrentStreams > 0)
        {
            CurrentStreams -= 1;
            Console.WriteLine("Поток остановлен");
        }
        else
        {
            Console.WriteLine("В данный момент не запущено ни одного потока");
        }
    }

    public void SetMaxStreams(int maxStreams)
    {
        MaxStreams = maxStreams;
    }

    public override void GetSubscriptionDetails()
    {
        base.GetSubscriptionDetails();
        Console.WriteLine($"Максимальное количество потоков: {MaxStreams}");
        Console.WriteLine($"Количество активных потоков: {CurrentStreams}");
        Console.WriteLine($"Возможность записи: {IsCanRecord}");
        Console.WriteLine($"Лимит устройств: {DeviceLimit}");
    }
}

class VideoSubscription : Subscription, IVideoSubscription
{
    public Dictionary<int, int> QualityCosts = new Dictionary<int, int> {{480, 0}, {720, 40}, {1080, 60}};
    private int _videoQuality = 480;
    public bool OfflineDownload {get; set;}
    public int VideoQuality
    {
        get => _videoQuality;
        private set
        {
            _videoQuality = value;
        }
    }
    private string IsOfflineDownload => OfflineDownload ? "Включена" : "Не включена";
    public int FinalCost => Cost + (QualityCosts.ContainsKey(VideoQuality) ? QualityCosts[VideoQuality] : 0);
    
    public VideoSubscription(int subscriptionId, string serviceName, int cost, int startedQuality, bool offlineDownload) : base(subscriptionId, serviceName, cost)
    {
        VideoQuality = startedQuality;
        OfflineDownload = offlineDownload;
    }

    public override void GetSubscriptionDetails()
    {
        base.GetSubscriptionDetails();
        Console.WriteLine($"Офлайн-загрузка: {IsOfflineDownload}");
        Console.WriteLine($"Цена с учетом повышения качества видео: {FinalCost}");
        Console.WriteLine($"Качество видео: {VideoQuality}p");
    }

    public override int CalculateMonthlyCost()
    {
        return FinalCost;
    }
    
    public void ChangeVideoQuality(int targetQuality)
    {
        if (QualityCosts.ContainsKey(targetQuality))
        {
            if (targetQuality > VideoQuality)
            {
                int priceDiff = QualityCosts[targetQuality] - QualityCosts[VideoQuality];
                if (Balance < priceDiff)
                {
                    Console.WriteLine("Недостаточно средств для изменения качества видео.");
                    return;
                }
                else
                {
                    Balance -= priceDiff;
                }   
            }
            VideoQuality = targetQuality;
            Console.WriteLine("Качество было успешно изменено.");       
        }
    }
}

class OnlineServiceSubscriptionForStaff : OnlineServiceSubscription
{
    public string Department { get; set; }
    public int EmployeeCount { get; set; }
    public bool CorporateAccess { get; set; }
    private string IsCorporateAccess => CorporateAccess ? "Включен" : "Не включен";
    public OnlineServiceSubscriptionForStaff(int subscriptionId, string serviceName, int currentUser, string department, int employeeCount, bool corporateAccess)
     : base(subscriptionId, serviceName, 0, currentUser, 1000, false)
    {
        Department = department;
        EmployeeCount = employeeCount;
        CorporateAccess = corporateAccess;
    }
    public override int CalculateMonthlyCost()
    {
        return Cost;
    }

    public override void GetSubscriptionDetails()
    {
        base.GetSubscriptionDetails();
        Console.WriteLine($"Отдел: {Department}");
        Console.WriteLine($"Количество сотрудников: {EmployeeCount}");
        Console.WriteLine($"Корпоративный доступ: {IsCorporateAccess}");
    }
    
}
```
