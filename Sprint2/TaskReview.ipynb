```
Employee employee = new Employee("Максим", "Стажер", 25000);
Manager manager = new Manager("Алексей", "Старший специалист", 60000, "IT-отдел");
ProjectManager projectManager = new ProjectManager("Георгий", "Тимлид", 120000, "IT-отдел", "Цифровизация ацтеков", [employee, manager]);

employee.PrintInfo();
Console.WriteLine("=========================================");
manager.PrintInfo();
Console.WriteLine("=========================================");
projectManager.PrintInfo();

projectManager.GenerateReport();
projectManager.GetMembers();

class Employee
{
    public string Name {get; protected set;}
    public string Position {get; protected set;}
    public float Salary {get; protected set;}

    public Employee(string name, string position, float salary)
    {
        Name = name;
        Position = position;
        Salary = salary;
    }

    public virtual void PrintInfo()
    {
        Console.WriteLine($"Имя: {Name}");
        Console.WriteLine($"Должность: {Position}");
        Console.WriteLine($"Зарплата: {Salary}");
    }
}

public interface IReportable
{
    void GenerateReport();
}

public interface ITeamLeader
{
    void GetMembers();
}

class Manager : Employee
{
    public string Department {get; protected set;}

    public Manager(string name, string position, float salary, string department) : base(name, position, salary)
    {
        Department = department;
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Отдел: {Department}");
    }
}

class ProjectManager : Manager, IReportable, ITeamLeader
{
    protected Employee[] managerMembers;
    public string ProjectName {get; protected set;}

    public ProjectManager(string name, string position, float salary, string department, string projectName, Employee[] members) : base(name, position, salary, department)
    {
        ProjectName = projectName;
        managerMembers = members;
    }

    public override void PrintInfo()
    {
        base.PrintInfo();
        Console.WriteLine($"Отвечает за проект: {ProjectName}");
    }

        public void GenerateReport()
    {
        Console.WriteLine($"Отчёт: Проект {ProjectName} выполнен раньше срока.");
    }

    public void GetMembers()
    {
        Console.WriteLine("Подчиненные: ");
        for (int i = 0; i < managerMembers.Length; i++)
        {
            Console.WriteLine(managerMembers[i].Name);
        }
    }
}

```
