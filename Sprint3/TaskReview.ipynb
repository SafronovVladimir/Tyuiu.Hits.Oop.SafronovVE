```

CPU cpu1 = new CPU("Intel Core i5", 15000);
CPU cpu2 = new CPU("Intel Core i5", 15000);
RAM ram = new RAM("Kingston 16GB", 5000);
GPU gpu = new GPU("RTX 4060", 35000);
HDD hdd = new HDD("WD Blue 1TB", 4000);

Console.WriteLine("=== Перегрузка операторов ===");
Console.WriteLine(cpu1 == cpu2);
Console.WriteLine(cpu1 != cpu2);
Console.WriteLine(cpu1.Equals(cpu2));
Console.WriteLine(cpu1.GetHashCode());

Console.WriteLine("\n=== ToString ===");
Console.WriteLine(cpu1.ToString());

Console.WriteLine("\n=== Интерфейс ===");
IInstallable installable = cpu1;
installable.Install();

Console.WriteLine("\n=== Полиморфизм ===");
Component[] components = { cpu1, ram, gpu, hdd };

foreach (Component component in components)
    component.GetInfo();

Console.WriteLine("\n=== Обобщённые методы ===");
Computer computer = new Computer();

foreach (Component component in components)
{
    if (component is CPU cpu)
        computer.AddComponent(cpu);
    else if (component is RAM r)
        computer.AddComponent(r);
    else if (component is GPU g)
        computer.AddComponent(g);
    else if (component is HDD h)
        computer.AddComponent(h);
}

computer.GetInfo();
Console.WriteLine($"Стоимость: {computer.GetTotalPrice()} руб.");

computer.RemoveComponent(hdd);
Console.WriteLine($"После удаления HDD: {computer.GetTotalPrice()} руб.");

Console.WriteLine("\n=== Фабрика компьютеров ===");

Computer[] configurations =
{
    ComputerFactory.CreateLowPriceConfig(),
    ComputerFactory.CreateMediumLowPriceConfig(),
    ComputerFactory.CreateMediumPriceConfig(),
    ComputerFactory.CreateHighPriceConfig(),
    ComputerFactory.CreatePremiumPriceConfig()
};

foreach (Computer pc in configurations)
{
    pc.GetInfo();
    Console.WriteLine($"Итого: {pc.GetTotalPrice()} руб.\n");
}

public interface IInstallable
{
    void Install();
}

public abstract class Component
{
    public string Model { get; set; }
    public decimal Price { get; set; }

    public Component(string model, decimal price)
    {
        Model = model;
        Price = price;
    }

    public virtual void GetInfo()
    {
        Console.WriteLine($"Модель: {Model}, Цена: {Price} руб.");
    }

    public override string ToString()
    {
        return $"{Model} - {Price} руб.";
    }
}

public class CPU : Component, IInstallable
{
    public CPU(string model, decimal price) : base(model, price)
    {
    }

    public override void GetInfo()
    {
        Console.WriteLine($"Процессор: {Model}, Цена: {Price} руб.");
    }

    public void Install()
    {
        Console.WriteLine($"CPU: {Model} установлен");
    }

    public static bool operator ==(CPU a, CPU b)
    {
        if (ReferenceEquals(a, b))
            return true;

        if (a is null || b is null)
            return false;

        return a.Model == b.Model;
    }

    public static bool operator !=(CPU a, CPU b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is CPU other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Model);
    }
}

public class RAM : Component, IInstallable
{
    public RAM(string model, decimal price) : base(model, price)
    {
    }

    public override void GetInfo()
    {
        Console.WriteLine($"Оперативная память: {Model}, Цена: {Price} руб.");
    }

    public void Install()
    {
        Console.WriteLine($"RAM: {Model} установлена");
    }

    public static bool operator ==(RAM a, RAM b)
    {
        if (ReferenceEquals(a, b))
            return true;

        if (a is null || b is null)
            return false;

        return a.Model == b.Model;
    }

    public static bool operator !=(RAM a, RAM b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is RAM other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Model);
    }
}

public class GPU : Component, IInstallable
{
    public GPU(string model, decimal price) : base(model, price)
    {
    }

    public override void GetInfo()
    {
        Console.WriteLine($"Видеокарта: {Model}, Цена: {Price} руб.");
    }

    public void Install()
    {
        Console.WriteLine($"GPU: {Model} установлена");
    }

    public static bool operator ==(GPU a, GPU b)
    {
        if (ReferenceEquals(a, b))
            return true;

        if (a is null || b is null)
            return false;

        return a.Model == b.Model;
    }

    public static bool operator !=(GPU a, GPU b)
    {
        return a != b;
    }

    public override bool Equals(object obj)
    {
        return obj is GPU other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Model);
    }
}

public class HDD : Component, IInstallable
{
    public HDD(string model, decimal price) : base(model, price)
    {
    }

    public override void GetInfo()
    {
        Console.WriteLine($"Жесткий диск: {Model}, Цена: {Price} руб.");
    }

    public void Install()
    {
        Console.WriteLine($"HDD: {Model} установлен");
    }

    public static bool operator ==(HDD a, HDD b)
    {
        if (ReferenceEquals(a, b))
            return true;

        if (a is null || b is null)
            return false;

        return a.Model == b.Model;
    }

    public static bool operator !=(HDD a, HDD b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is HDD other && this == other;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Model);
    }
}

public class Computer
{
    private List<Component> components = new List<Component>();

    public void AddComponent<T>(T component) where T : Component, IInstallable
    {
        components.Add(component);
        component.Install();
    }

    public void RemoveComponent<T>(T component) where T : Component
    {
        components.Remove(component);
    }

    public decimal GetTotalPrice()
    {
        decimal total = 0;

        foreach (Component component in components)
        {
            total += component.Price;
        }

        return total;
    }

    public void GetInfo()
    {
        Console.WriteLine("Комплектующие компьютера:");

        foreach (Component component in components)
        {
            component.GetInfo();
        }
    }
}

public static class ComputerFactory
{
    public static Computer CreateLowPriceConfig()
    {
        Computer computer = new Computer();
        computer.AddComponent(new CPU("XEON E6580 Б/У", 3000));
        computer.AddComponent(new GPU("AMD RADEON RX570 Б/У", 3500));
        computer.AddComponent(new RAM("Оперативная память Chinases 16GB DDR3 1800 Mhz Б/У", 3400));
        computer.AddComponent(new HDD("480GB Seagate Low-Profile Б/У", 2200));

        return computer;
    }

    public static Computer CreateMediumLowPriceConfig()
    {
        Computer computer = new Computer();
        computer.AddComponent(new CPU("I3 12400F", 10000));
        computer.AddComponent(new GPU("AMD RADEON RX 6500", 16000));
        computer.AddComponent(new RAM("Оперативная память Kingstone 16GB DDR4 3600 Mhz", 6000));
        computer.AddComponent(new HDD("HDD 1TB WD BLUE", 5200));

        return computer;
    }


    public static Computer CreateMediumPriceConfig()
    {
        Computer computer = new Computer();
        computer.AddComponent(new CPU("Intel Core i5-14400F", 18000));
        computer.AddComponent(new GPU("NVIDIA GeForce RTX 4060 8GB", 32000));
        computer.AddComponent(new RAM("Kingston Fury Beast 32GB DDR5 5600 MHz", 12000));
        computer.AddComponent(new HDD("SSD Samsung 990 EVO 1TB", 10000));

        return computer;
    }

    public static Computer CreateHighPriceConfig()
    {
        Computer computer = new Computer();
        computer.AddComponent(new CPU("AMD Ryzen 7 7800X3D", 35000));
        computer.AddComponent(new GPU("NVIDIA GeForce RTX 4070 Ti SUPER 16GB", 85000));
        computer.AddComponent(new RAM("Corsair Vengeance 32GB DDR5 6000 MHz", 16000));
        computer.AddComponent(new HDD("SSD Samsung 990 PRO 2TB", 20000));

        return computer;
    }

    public static Computer CreatePremiumPriceConfig()
    {
        Computer computer = new Computer();
        computer.AddComponent(new CPU("AMD Ryzen 9 9950X3D", 80000));
        computer.AddComponent(new GPU("NVIDIA GeForce RTX 5090 32GB", 300000));
        computer.AddComponent(new RAM("G.Skill Trident Z5 64GB DDR5 6400 MHz", 35000));
        computer.AddComponent(new HDD("SSD Samsung 9100 PRO 4TB", 55000));

        return computer;
    }

}
```
