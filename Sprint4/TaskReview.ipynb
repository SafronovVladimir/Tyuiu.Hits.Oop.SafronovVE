```
Enterprise bigEnterpise = new Enterprise();
EnterpriseService bigEnterpiseService = new EnterpriseService(bigEnterpise);

Supplier chairSupplier = new Supplier("ДваСтула", 12000, "Поставщик стульев");
Supplier networkSupplier = new Supplier("ИнтернетБезИнтернета", 20000, "Провайдер");
Supplier happinessSupplier = new Supplier("Счастье", 0, "Поставщик счастья");

Employee employee1 = new Employee("Михаил", "Прораб", 100000);
Employee employee2 = new Employee("Владислав", "Стажер", 25000);
Employee employee3 = new Employee("Данте", "HR", 50000);

bigEnterpise.AddSupplier(chairSupplier);
bigEnterpise.AddSupplier(networkSupplier);
bigEnterpise.AddSupplier(happinessSupplier);

bigEnterpise.AddEmployee(employee1);
bigEnterpise.AddEmployee(employee2);
bigEnterpise.AddEmployee(employee3);

bigEnterpiseService.DisplaySupplier();
Console.WriteLine();
bigEnterpiseService.DisplayEmployees();
Console.WriteLine();
bigEnterpiseService.DisplayCalculatedTotalPrice();

public interface IEnterprise
{
    void AddSupplier(ISupplier supplier);
    void AddEmployee(IEmployee employee);
    List <ISupplier> GetSuppliers();
    List <IEmployee> GetEmployees();
    decimal CalculateTotalExpenses();
}

public interface ISupplier
{
    string GetName();
    decimal GetCost();
    string SupplierType {get; set;}
}

public interface IEmployee
{
    string GetFullName();
    decimal GetSalary();
    string GetPosition();
    string Position {get; set;}
}

public class Enterprise : IEnterprise
{
    List<ISupplier> suppliers = new List<ISupplier>();
    List<IEmployee> employees = new List<IEmployee>();

    public void AddSupplier(ISupplier supplier)
    {
        suppliers.Add(supplier);
    }
    
    public void AddEmployee(IEmployee employee)
    {
        employees.Add(employee);
    }

    public List<ISupplier> GetSuppliers()
    {
        return suppliers;
    }

    public List<IEmployee> GetEmployees()
    {
        return employees;
    }
    
    public decimal CalculateTotalExpenses()
    {
        decimal total = 0;

        foreach (var supplier in suppliers)
        {
            total += supplier.GetCost();
        }

        foreach (var employee in employees)
        {
            total += employee.GetSalary();
        }

        return total;
    }
}

public class Supplier : ISupplier
{
    string Name {get; set;}
    decimal Cost {get; set;}
    public string SupplierType {get; set;}
    
    public Supplier(string name, int cost, string supplierType)
    {
        Name = name;
        Cost = cost;
        SupplierType = supplierType;
    }

    public string GetName()
    {
        return Name;
    }

    public decimal GetCost()
    {
        return Cost;
    }
}

public class Employee : IEmployee
{
    string Name {get; set;}
    public string Position {get; set;}
    decimal Salary {get; set;}

    public Employee(string name, string position, decimal salary)
    {
        Name = name;
        Position = position;
        Salary = salary;
    }

    public decimal GetSalary()
    {
        return Salary;
    }

    public string GetFullName()
    {
        return Name;
    }

    public string GetPosition()
    {
        return Position;
    }
}

class EnterpriseService
{
    private readonly IEnterprise _enterpise;
    public EnterpriseService(IEnterprise enterprise)
    {
       _enterpise = enterprise; 
    }

    public void DisplaySupplier()
    {
        Console.WriteLine("Список поставщиков: ");
        int k = 1;
        foreach (var supplier in _enterpise.GetSuppliers())
        {
            Console.WriteLine($"{k}. {supplier.GetName()} - Стоимость услуг: {supplier.GetCost()}");
            k++;
        }
    }

    public void DisplayEmployees()
    {
        Console.WriteLine("Список сотрудников:");
        int k = 1;
        foreach (var employee in _enterpise.GetEmployees())
        {
            Console.WriteLine($"{k}. {employee.GetFullName()} - Должность: {employee.GetPosition()} - Зарплата: {employee.GetSalary()}");
            k++;
        }
    }

    public void DisplayCalculatedTotalPrice()
    {
        Console.WriteLine($"Общие расходы: {_enterpise.CalculateTotalExpenses()}");
    }
}
```
